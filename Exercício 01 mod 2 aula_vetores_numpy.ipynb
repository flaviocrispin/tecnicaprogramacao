```python
classes = ['pet', 'tecnologia', 'alimentação', 'esportes']
```

```python
bancos_dados = {
    "cachorro": np.array([0.9, 0.0, 0.0, 0.0]),
    "gato": np.array([0.9, 0.0, 0.0, 0.0]),
    "passarinho": np.array([0.9, 0.0, 0.0, 0.0]),

    "computador": np.array([0.0, 0.9, 0.0, 0.0]),
    "robo": np.array([0.0, 0.9, 0.0, 0.0]),

    "arroz": np.array([0.0, 0.0, 0.9, 0.0]),
    "carne": np.array([0.0, 0.0, 0.9, 0.0]),

    # Nova categoria: esportes
    "futebol": np.array([0.0, 0.0, 0.0, 0.9]),
    "basquete": np.array([0.0, 0.0, 0.0, 0.9]),
    "tenis": np.array([0.0, 0.0, 0.0, 0.9]),
    }
```

```python
pergunta = "Eu gosto de futebol"
```

```python
tokens = pergunta.lower().split()
```

```python
tokens
```

```python
vetores = [bancos_dados[token] for token in tokens if token in bancos_dados]
```

```python
vetores
```

```python
palavras_encontradas = [token for token in tokens if token in bancos_dados]
```

```python
palavras_encontradas
```

```python
vetor_media = np.mean(vetores, axis=0)
```

```python
vetor_media
```

```python
indice_vencedor = np.argmax(vetor_media)
```

```python
indice_vencedor
```

```python
tema = classes[indice_vencedor]
tema
```

```python
print (f"A pergunta '{pergunta}")
print (f"a palavra encontrada foi: '{palavras_encontradas}'")
print (f"O Tema da pergunta é: '{tema}'")
```
