```python
# DESAFIO 1 - Expansão de Domínio

import numpy as np

classes = ['pet', 'tecnologia', 'alimentação', 'finanças']

bancos_dados = {
    "cachorro": np.array([0.0, 0.2, 0.9, 0.0]),
    "computador": np.array([0.0, 0.9, 0.1, 0.0]),
    "pizza": np.array([0.0, 0.1, 0.9, 0.0]),
    "dinheiro": np.array([0.0, 0.0, 0.1, 0.9]),
    "banco": np.array([0.0, 0.0, 0.2, 0.8]),
    "investimento": np.array([0.0, 0.0, 0.1, 0.9])
}

pergunta = "Quero fazer um investimento"

palavras = pergunta.lower().split()

vetores = []

for palavra in palavras:
    if palavra in bancos_dados:
        vetores.append(bancos_dados[palavra])

if vetores:
    vetor_medio = np.mean(vetores, axis=0)
    indice = np.argmax(vetor_medio)
    print("Tema:", classes[indice])
else:
    print("Nenhuma palavra conhecida foi encontrada na base de dados")
```

```python
# DESAFIO 2 - Frases Mistas

pergunta = "O computador quebrou"

palavras = pergunta.lower().split()

vetores = []

for palavra in palavras:
    if palavra in bancos_dados:
        vetores.append(bancos_dados[palavra])

if vetores:
    vetor_medio = np.mean(vetores, axis=0)
    indice = np.argmax(vetor_medio)

    print("Vetores encontrados:", vetores)
    print("Média dos vetores:", vetor_medio)
    print("Tema:", classes[indice])
else:
    print("Nenhuma palavra conhecida foi encontrada na base de dados")
```
