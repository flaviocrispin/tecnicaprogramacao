```python
import numpy as np

classes = ['pet', 'tecnologia', 'alimentação', 'esportes']

bancos_dados = {
    "cachorro": np.array([0.0, 0.2, 0.9, 0.0]),
    "gato": np.array([0.9, 0.0, 0.2, 0.0]),
    "passarinho": np.array([0.9, 0.0, 0.2, 0.0]),

    "robo": np.array([0.1, 0.9, 0.0, 0.0]),
    "computador": np.array([0.0, 0.9, 0.0, 0.0]),
    "dados": np.array([0.0, 0.9, 0.0, 0.0]),

    "arroz": np.array([0.0, 0.0, 0.9, 0.0]),
    "carne": np.array([0.0, 0.0, 0.9, 0.0]),
    "racao": np.array([0.0, 0.0, 0.9, 0.0]),

    "natação": np.array([0.0, 0.0, 0.0, 0.9]),
    "basquete": np.array([0.0, 0.0, 0.0, 0.9]),
    "corrida": np.array([0.0, 0.0, 0.0, 0.9]),
}
```

```python
pergunta = "a natação é divertida?"
tokens = pergunta.lower().replace("?", "").split()
vetores = [bancos_dados[token] for token in tokens if token in bancos_dados]

if vetores:
    vetor_media = np.mean(vetores, axis=0)
    indice_vencedor = np.argmax(vetor_media)
    tema = classes[indice_vencedor]
    print(f"Tema identificado: {tema}")
else:
    print("Nenhuma palavra conhecida foi encontrada na base de dados.")
```
