```python

# Exercício 1 - Expansão de Domínio (Novas Categorias) 

import numpy as np

classes = ['pet', 'tecnologia', 'alimentação', 'esportes']

bancos_dados = {
"cachorro": np.array([0.9, 0.0, 0.0, 0.0]),
"gato": np.array([0.9, 0.0, 0.0, 0.0]),
"passarinho": np.array([0.9, 0.0, 0.0, 0.0]),

"robo": np.array([0.0, 0.9, 0.0, 0.0]),
"computador": np.array([0.0, 0.9, 0.0, 0.0]),
"dados": np.array([0.0, 0.9, 0.0, 0.0]),

"arroz": np.array([0.0, 0.0, 0.9, 0.0]),
"carne": np.array([0.0, 0.0, 0.9, 0.0]),
"racao": np.array([0.0, 0.0, 0.9, 0.0]),

"futebol": np.array([0.0, 0.0, 0.0, 0.9]),
"basquete": np.array([0.0, 0.0, 0.0, 0.9]),
"corrida": np.array([0.0, 0.0, 0.0, 0.9]),
}

```

```python
# Exercício 2 - Frases Mistas e Múltiplas Palavras

pergunta = "O computador quebrou"

tokens = pergunta.lower().split()

vetores = [bancos_dados[token] for token in tokens if token in bancos_dados]

print(vetores)
```

```python
pergunta = "O cachorro derrubou o computador"

tokens = pergunta.lower().split()

vetores = [bancos_dados[token] for token in tokens if token in bancos_dados]

print(vetores)
```

```python
np.mean(vetores, axis=0)
```

```python
# Exercício 3 - Tratamento de Palavras Desconhecidas (Robusto contra Erros) 

pergunta = "o cometa é brilhante"

tokens = pergunta.lower().split()

vetores = [bancos_dados[token] for token in tokens if token in bancos_dados]

if len(vetores) > 0:
    media = np.mean(vetores, axis=0)

    indice_classe = np.argmax(media)
    classe_prevista = classes[indice_classe]

    print("Classe identificada:", classe_prevista)

else:
    
    print("Nenhuma palavra conhecida foi encontrada na base de dados.")
```
